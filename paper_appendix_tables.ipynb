```python
import pandas as pd

cust_metrics_df = pd.read_csv("pfairness_questions - custom metric definition.csv")
```

```python
def format_default(val):
    if isinstance(val, float):
        return f"{round(val, 2)}"
    return str(val)

def format_to_percentage(val):
    if isinstance(val, float):
        return f"{round(val * 100, 1)}%"
    return str(val)

def df_to_latex(df, format="default"):
    formatted_df = df.copy()
    for col in formatted_df.columns:
        if format == "default":
            formatfunc = format_default
        elif(format == "percentage"):
            formatfunc = format_to_percentage
        formatted_df[col] = formatted_df[col].apply(formatfunc)
    
    results_str = formatted_df.to_latex(index=False, escape=True)
    
    # patch correction:
    results_str = results_str.replace("70b", "70B").replace("72b", "72B").replace("qwen25", "Qwen25")
    results_str = results_str.replace(r"employment\_sector", "industry").replace(r"nationality\_group", "nationality")
    
    return results_str

def df_to_latex_widetable(df, caption, format="default"):
    latex_body = df.to_latex(index=False, 
                             escape=True,
                             column_format='l p{0.75\\columnwidth}',
                             longtable=False)

    return r"""\begin{table}[H]
\resizebox{\columnwidth}{!}{%
""" + latex_body + rf"""}}
\caption{{{caption.replace("_",r"\_")}}}
\label{{tab:{caption}}}
\end{{table}}
"""

def df_to_latex_one_column(df, caption, label, format="default"):
    return r"""\begin{table}[H]
\resizebox{\columnwidth}{!}{%
""" + df_to_latex(df, format) + rf"""}}
\caption{{{caption.replace("_",r"\_")}}}
\label{{tab:{label}}}
\end{{table}}
"""
```

```python
print(df_to_latex_widetable(cust_metrics_df[:3], caption="asdasd"))
print()
print(df_to_latex_widetable(cust_metrics_df[3:9], caption="asdasd"))
print()
print(df_to_latex_widetable(cust_metrics_df[9:], caption="asdasd"))
```
